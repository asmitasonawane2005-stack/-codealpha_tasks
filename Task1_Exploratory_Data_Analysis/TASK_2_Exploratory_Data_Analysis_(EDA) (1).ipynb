```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel("samplesuperstoredata.xlsx")
df.head()
```

```python
df.columns
```

```python
df.info()
df.describe()
```

Remove spaces in column names

```python
df.columns = df.columns.str.strip()
```

Check missing values

```python
df.isnull().sum()
```

Remove missing values

```python
df.dropna(inplace=True)
```

Remove duplicates

```python
df.drop_duplicates(inplace=True)
```

# Question 1:
Which category has the highest total sales?

```python
df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
```

Result:
Technology category has the highest sales.

# Question 2:
Which region generates the maximum profit?

```python
df.groupby('Region')['Profit'].sum().sort_values(ascending=False)
```

Result:
West region has the highest profit.

# Question 3:
What are the top 5 products by sales?

```python
df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head()
```

Result:
Top 5 products generate the highest sales are as above

# Question 4:
Are there any products making a loss?

```python
df[df['Profit'] < 0][['Product Name', 'Profit']]
```

Result:
 Yes..Some products are making losses.

# Question 5:
What is the relationship between sales and profit?

```python
df.groupby('Category')['Sales'].sum().plot(kind='bar')
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.show()
```

Result:
Sales and profit are positively related.
